```python
import os
import requests
import pymongo
import json
#from dotenv import load_dotenv

# Load environment variables
#load_dotenv()

def get_ebay_credentials():
    """Fetch eBay API credentials from environment variables."""
    return {
        "app_id": os.getenv("EBAY_APP_ID"),
        "cert_id": os.getenv("EBAY_CERT_ID"),
        "dev_id": os.getenv("EBAY_DEV_ID"),
        "auth_token": os.getenv("EBAY_OAUTH_TOKEN"),
        "finding_api": os.getenv("EBAY_FINDING_API"),
    }

# Fetch credentials
creds = get_ebay_credentials()
print("eBay Credentials Loaded Successfully")
# Include the database name in the URI
MONGO_URI = "mongodb+srv://mosesmacharia084:Newpassword18702@cluster0.gcck2ys.mongodb.net/ebay_products?retryWrites=true&w=majority"  
client = pymongo.MongoClient(MONGO_URI)
# Access the database (this will create it if it doesn't exist)
db = client["ebay_products"] 
print("Connected to MongoDB")



def fetch_ebay_products(keyword):
    """Fetch eBay product listings using Finding API."""
    url = creds["finding_api"]
    
    params = {
        "OPERATION-NAME": "findItemsByKeywords",
        "SERVICE-VERSION": "1.0.0",
        "SECURITY-APPNAME": creds["app_id"],
        "RESPONSE-DATA-FORMAT": "JSON",
        "REST-PAYLOAD": "",
        "keywords": keyword,
        "paginationInput.entriesPerPage": 10
    }
    
    response = requests.get(url, params=params)
    
    if response.status_code == 200:
        return response.json().get("findItemsByKeywordsResponse", [])[0].get("searchResult", [])[0].get("item", [])
    else:
        return {"error": "Failed to fetch data", "status_code": response.status_code}

# Test API Request
keyword = "asus rog"
results = fetch_ebay_products(keyword)

# Pretty Print Results
print(json.dumps(results, indent=4))

# Store in MongoDB
if results:
    db["products"].insert_one({"search_query": keyword, "results": results})
    print("Data Stored in MongoDB Successfully")
```

```python
import pymongo

# MongoDB Connection
MONGO_URI = "mongodb+srv://mosesmacharia084:Newpassword18702@cluster0.gcck2ys.mongodb.net/ebay_products?retryWrites=true&w=majority"
client = pymongo.MongoClient(MONGO_URI)
db = client["ebay_products"]
collection = db["products"]

def fetch_stored_products(search_query):
    """Retrieve products from MongoDB and format the output."""
    stored_data = collection.find_one({"search_query": search_query})

    if not stored_data or "results" not in stored_data:
        return "No results found for this search query."

    formatted_output = []
    
    for idx, product in enumerate(stored_data["results"], 1):
        title = product.get("title", "No Title")
        
        # Safely extract price details
        selling_status = product.get("sellingStatus", [{}])  # Default to [{}] to avoid index errors
        
        # Check if convertedCurrentPrice is a list and get the first element
        current_price_data = selling_status[0].get("convertedCurrentPrice", [{}])
        if isinstance(current_price_data, list) and current_price_data:
            current_price_data = current_price_data[0] 
        else:
            current_price_data = {}  # Set to an empty dictionary if not a list or empty

        # Extract price and currency safely
        price = current_price_data.get("__value__", "N/A")
        currency = current_price_data.get("@currencyId", "N/A")

        url = product.get("viewItemURL", "No URL")
        
        formatted_output.append(f"{idx}. {title}\n   Price: {price} {currency}\n   Link: {url}\n")
    
    return "\n".join(formatted_output)

# Example Usage
search_keyword = "asus rog"  # Change this to match a stored query
output = fetch_stored_products(search_keyword)
print(output)
```
